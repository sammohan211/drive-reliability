```python
import duckdb
```

```python
con = duckdb.connect()
```

```python
con.sql("CREATE VIEW drives AS SELECT * FROM read_parquet('../data/parquet/*.parquet', union_by_name=true)")
```

```python
con.sql("DESCRIBE drives")
```

```python
con.sql("SUMMARIZE SELECT * FROM '../data/parquet/data_Q1_2024.parquet'")
```

```python
con.sql("SELECT model, count(DISTINCT serial_number) n FROM drives GROUP BY 1 ORDER BY n DESC LIMIT 20")
```

```python
con.sql("SELECT * FROM drives USING SAMPLE 1000").df()
```
